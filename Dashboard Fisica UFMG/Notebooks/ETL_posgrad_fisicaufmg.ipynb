```python
import pandas as pd
import locale
locale.setlocale(locale.LC_TIME, 'pt_BR.UTF-8')
```

```python
# Link para extrair as tabelas das tags htlm
source_data: str = "https://www.fisica.ufmg.br/posgraduacao/corpo-discente/"

# Carregando os dados
mest, doc = pd.read_html(source_data, header=1)
mest["Modalidade"] = "Mestrado"
doc["Modalidade"] = "Doutorado" 

# Linha "morta" da tabela
mest = mest.head(-1) 

pos_grad = pd.concat([mest, doc], axis=0, ignore_index=True)     # DataFrame único com as duas tabelas
pos_grad = pos_grad.drop(columns="Matrícula") # Removendo coluna desnecessária 

pos_grad.sample(5)
```

## <font color="#5e5e5e">Tratamento dos Dados</font>

__Renomeando Colunas__


```python
print("Colunas Iniciais do Dataset:\n")
pos_grad.head(0)
```

```python
# Renomeando colunas 
pos_grad.columns = pos_grad.columns.str.replace(" ", "_").str.lower()
print(pos_grad.columns.tolist())

# Removendo caracteres ( ´ , ç , ^)
cols_rename: dict[str, str] = {
    "aluno/a": "aluno",
    "entrada": "inicio_curso",
    "área_de_concentração": "area",
    "orientador/a": "orientador",
    "agência": "agencia",
    "início_bolsa": "inicio_bolsa",
    "término_bolsa": "termino_bolsa",
    "prazo_término_curso": "termino_curso",
}
pos_grad = pos_grad.rename(columns=cols_rename)
print("Colunas renomeadas:\n")
pos_grad.head(0)
```

```python
# Reordenamento das colunas
ordem_cols = ["aluno", "modalidade", "area", "orientador", "inicio_curso", "termino_curso","bolsista","agencia", "inicio_bolsa", "termino_bolsa", ]
pos_grad = pos_grad[ordem_cols]
pos_grad
```

__Tipos de dados__

```python
pos_grad.dtypes
```

**Datas**

```python
# Formatação de datas: mês/ano (jan/XX)
cols_datetime = ["inicio_curso", "inicio_bolsa", "termino_bolsa", "termino_curso"] 

for col in cols_datetime:
    pos_grad[col] = pd.to_datetime(pos_grad[col], format="%b/%y")
pos_grad.dtypes
```

**Booleanos**

```python
# Convertendo bolsista para booleano
print(f"Valores únicos de bolsista (antes)\n{pos_grad["bolsista"].unique()}")

map_bolsista = {"sim": True, "não": False}
pos_grad["bolsista"] = pos_grad["bolsista"].str.lower().map(map_bolsista)

print(f"Valores únicos de bolsista (depois)\n{pos_grad["bolsista"].unique()}")
pos_grad.dtypes
```

__Tratamento de Strings__

```python
print(f"Agências de Fomento (antes): {pos_grad["agencia"].unique()}")

# Correção do typo CNPq/PRPG e padronização em maiúscula
pos_grad["agencia"] = pos_grad["agencia"].str.upper()

print(f"Agências de Fomento (depois): {pos_grad["agencia"].unique()}")
```

```python
# Formatando nome de pessoas com "Nome Sobrenome Sobrenome2" 
for col in ["aluno", "orientador"]:
    pos_grad[col] = pos_grad[col].str.title().str.strip()    
```

**Padronização da Coluna de Orientadores**

Algumas colunas possuem informação, além do orientador, também da co-orientação do aluno. Manterei **somente o orientador.**

```python
pos_grad.orientador.unique()
```

```python
re_parentese = r"\s*\(.*?\)"
pos_grad.orientador= pos_grad.orientador.str.replace(re_parentese, repl="", regex=True).str.strip()
print(pos_grad.orientador.unique())
```

```python
pos_grad.orientador.sort_values().unique()
# Ado Jorio, Carlos Basilio, Glauber Dorsch, Karla Balzuweit, Livia Siman (3), Maria Carolina
```

Como estratégia, vou deixar apenas nome e sobrenome dos orientadores.

```python
def separa_nome_sobrenome(nome_completo: str) -> str:
    """Padronizar os nomes do DataFrame em nome e sobrenome"""
    nomes: list[str] = nome_completo.split()[:2]
    nome, sobrenome = nomes
    novo_nome =  nome + " " + sobrenome
    return novo_nome
```

```python
pos_grad.orientador = pos_grad.orientador.apply(separa_nome_sobrenome)
```

```python
pos_grad.orientador.sort_values().unique()
```

Remoção de acentuações, corrigirá alguns casos como *Livia Siman* e *Carlos Basílio*

```python
pos_grad.orientador = pos_grad.orientador.str.replace("í","i").str.replace("á", "a")
```

```python
pos_grad.orientador.sort_values().unique()
```

Precisamos dar conta de:
- Nomes Compostos: *Guilherme Da* e *Sebastião De*
- Typos: Nelson Yokomizo $\times$ Nelson Yokomiso

```python
mask_sebastiao = (pos_grad["orientador"] == "Sebastião De")
mask_guilherme = (pos_grad["orientador"] == "Guilherme Da")

pos_grad.loc[mask_sebastiao, "orientador"] = "Sebastião de Pádua"
pos_grad.loc[mask_guilherme, "orientador"] = "Guilherme Da Silva"

pos_grad.orientador = pos_grad.orientador.str.replace("Nelson Yokomiso", "Nelson Yokomizo")
```
